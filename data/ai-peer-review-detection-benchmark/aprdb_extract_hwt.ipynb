# Export for finetuning as jsonl file

```python
import pandas as pd
import json
import random

def create_dataset(df, filename="/data2/cyaiche/train_human_style.jsonl"):
    # Templates d'instructions en anglais pour la diversité
    prompts = [
        "Write a scientific contribution for the {conf} conference, as expected in {year}.",
        "Generate a research statement in the style of {conf} ({year}).",
        "Compose a scholarly text suitable for {conf} based on the standards of {year}.",
        "Produce an academic abstract/content for {conf}, year {year}."
    ]

    with open(filename, 'w', encoding='utf-8') as f:
        for _, row in df.iterrows():
            # Sélection d'un prompt aléatoire
            instruction = random.choice(prompts).format(conf=row['conference'], year=row['year'])

            payload = {
                "messages": [
                    {
                        "role": "system",
                        "content": "You are a professional academic researcher. Your writing is concise, uses diverse vocabulary, and follows human-like syntactic structures with rich adverbial usage."
                    },
                    {
                        "role": "user",
                        "content": instruction
                    },
                    {
                        "role": "assistant",
                        "content": str(row['content'])
                    }
                ]
            }
            f.write(json.dumps(payload) + '\n')

# Utilisation : create_english_dataset(df)
```
