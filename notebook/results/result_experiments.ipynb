```python
def build_sage_long(df_sage, hash_cible):
    # Sélection de la ligne cible (référence par hash)
    row = df_sage.loc[df_sage["hash"] == hash_cible]

    # Garde-fous : existence + unicité du hash
    if len(row) == 0:
        raise ValueError(f"hash introuvable : {hash_cible}")
    if len(row) > 1:
        raise ValueError(f"hash non unique ({len(row)} lignes) : {hash_cible}")

    # Cohérence des 3 colonnes-listes parallèles avant explosion
    n = len(row.iloc[0]["ids"])
    for col in ("ids", "content", "y_true", "y_scores"):
        if len(row.iloc[0][col]) != n:
            raise ValueError(
                f"longueur incohérente : {col}={len(row.iloc[0][col])} vs ids={n}"
            )

    # Explosion parallèle : 1 ligne = 1 review-id, y_true/y_scores alignés positionnellement
    # position = indice k dans les listes d'origine (trace l'ordre de ids)
    sage_long = (
        row.assign(position=[list(range(n))])
           .explode(["ids", "y_true", "y_scores", "position", "content"])
           .rename(columns={"ids": "review-id"})
           .reset_index(drop=True)
    )
    return sage_long
```

```python
def attach_dataset(sage_long, df_dataset):
    # Jointure sur le couple (review-id, content) : clé désormais unique à droite
    merged = sage_long.merge(
        df_dataset,
        on=["review-id", "content"],
        how="left",
        validate="many_to_one",
    )

    # Garde-fou : aucun couple orphelin. On teste sur une colonne exclusive
    # à df_dataset (pas content, qui vient déjà de gauche et ne sera jamais NaN ici).
    col_temoin = [c for c in df_dataset.columns
                  if c not in ("review-id", "content")][0]
    manquants = merged.loc[merged[col_temoin].isna(), ["review-id", "content"]]
    if len(manquants):
        raise ValueError(
            f"{len(manquants)} couple(s) (review-id, content) absent(s) de df_dataset :\n{manquants}"
        )

    return merged
```

```python
def attach_sentences(merged, df_res):
    # Matérialise l'ordre séquentiel des phrases (ordre implicite des lignes)
    df_res_ord = df_res.reset_index(drop=True).reset_index(names="sent_order")

    final = merged.merge(
        df_res_ord,
        on="content",
        how="left",
        validate="one_to_many",   # 1 content -> N phrases
    )

    # Garde-fou : tout content doit avoir au moins une phrase dans df_res
    sans_phrase = final.loc[final["sent_order"].isna(), "content"]
    if len(sans_phrase):
        raise ValueError(
            f"{sans_phrase.nunique()} content(s) sans phrase dans df_res"
        )

    # Tri final : d'abord l'ordre des reviews (position),
    # puis l'ordre des phrases au sein du review (sent_order)
    final = (
        final.sort_values(["position", "sent_order"])
             .reset_index(drop=True)
    )
    return final
```
