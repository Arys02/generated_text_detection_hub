```python
import pandas as pd
```

```python
import json
from src.config import RESULT_DIR

path = RESULT_DIR / 'results.jsonl'
with path.open(encoding='utf-8') as f:
    for lineno, line in enumerate(f, start=1):
        s = line.strip()
        if not s:
            continue
        try:
            json.loads(s)
        except json.JSONDecodeError as e:
            print(f"ligne {lineno} | {e.msg} @ col {e.pos} | len={len(s)}")
            print(repr(s[max(0, e.pos - 80):e.pos + 80]))
```
