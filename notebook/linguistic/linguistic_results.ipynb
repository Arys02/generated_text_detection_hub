```python
from src.config import RESULT_DIR
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import RocCurveDisplay

local_dir = RESULT_DIR / "implicite_segmentation"
gru = "gru_result_1.npz"
mlp = "mlp_result_1.npz"
d_gru = np.load(local_dir / gru)
d_mlp = np.load(local_dir / mlp)
y_gru_true, y_gru_proba = d_gru["y_true"], d_gru["y_proba"]
y_mlp_true, y_mlp_proba = d_mlp["y_true"], d_mlp["y_proba"]


fig, ax = plt.subplots(figsize=(6, 6))
RocCurveDisplay.from_predictions(y_gru_true, y_gru_proba, ax=ax, name="GRU")
RocCurveDisplay.from_predictions(y_mlp_true, y_mlp_proba, ax=ax, name="MLP baseline")
ax.plot([0, 1], [0, 1], "--", color="grey", label="aléatoire")
ax.legend(); plt.tight_layout(); plt.show()
```
