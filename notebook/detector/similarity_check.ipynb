```python
def contrast_score(sim, target):
    generators = [g for g in sim.keys() if g != target]
    contrast_scores = {}
    for i in generators:
        others = [j for j in generators if j != i]
        mean_intra_ai = sum(sim[i][j] for j in others) / len(others)
        sim_to_human = sim[i][target]
        contrast_scores[i] = mean_intra_ai - sim_to_human

    for g, s in sorted(contrast_scores.items(), key=lambda x: -x[1]):
        print(f"{g:10s} {s:+.4f}")
    return contrast_scores
```

```python
import json
from src.config import DATA_DIR
from collections import defaultdict

with open(DATA_DIR / "similarities_2.json") as f:
    data = json.load(f)

# Matrice symétrique sim[i][j]
sim = defaultdict(dict)
for row in data:
    a, b, v = row["a"], row["b"], row["value"]
    sim[a][b] = v
    sim[b][a] = v

cs = contrast_score(sim, "human")
for g, s in sorted(cs.items(), key=lambda x: -x[1]):
    print(f"{g:10s} {s:+.4f}")
```
